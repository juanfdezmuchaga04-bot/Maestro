# Maestro — free NVIDIA GPU node

This notebook runs **your own Maestro** on the free NVIDIA GPU of this session and connects it to Maestro on your computer, so renders use Maestro's real engine exactly as on an NVIDIA PC.

**Before running** (Kaggle): *Settings → Accelerator → GPU T4 ×2* and *Settings → Internet → On* (both need a phone-verified account). On Lightning AI, open a Studio with a T4 and run the same cells in a terminal or notebook.

Then *Run all*. The last cell prints a `maestro-node:…` code: paste it in Maestro → **Auto → Free NVIDIA GPU → Connect**. Keep this tab open while you render; the free quota (≈30 GPU-hours/week on Kaggle) is consumed while the session runs. Stop the session when you finish.

Optional secrets (*Add-ons → Secrets*): `HF_TOKEN` (gated models), `GITHUB_TOKEN` (only if your Maestro repository is private).

```python
# 1. Get Maestro (change REPO/BRANCH if you use your own fork)
REPO = "https://github.com/juanfdezmuchaga04-bot/Maestro.git"
BRANCH = "main"
import os, subprocess
def secret(name):
    try:
        from kaggle_secrets import UserSecretsClient
        return UserSecretsClient().get_secret(name)
    except Exception:
        return os.environ.get(name, "")
token = secret("GITHUB_TOKEN")
url = REPO.replace("https://", f"https://x-access-token:{token}@") if token else REPO
if os.environ.get("HF_TOKEN") is None and secret("HF_TOKEN"):
    os.environ["HF_TOKEN"] = secret("HF_TOKEN")
if not os.path.isdir("Maestro"):
    subprocess.run(["git", "clone", "--depth", "1", "--branch", BRANCH, url, "Maestro"], check=True)
else:
    subprocess.run(["git", "-C", "Maestro", "pull", "--ff-only"], check=True)
print("Maestro checked out")
```
